```python
import pyaraas

from pyaraas import Transform, Vector
from pyaraas import Task

workcell = pyaraas.start("PandaRL")

target = workcell.get_peripheral("insertion_box-0")
target.set_collision_model(True, True)
```

```python
p = workcell.add_part("peg", "peg0", Transform(560,50,350,0,-0,0))
xform = p.get_grasp("up", 0)
print(xform[0])
goal = p.get_xform().multiply(xform[0])
goal.position = goal.multiply(pyaraas.Vector(0,0,50))
print(goal)
```

```python
g = workcell.get_gripper("PdDefault-0")
await g.move_to(goal, 0.1)
await g.open(60, 0.1)
goal.position = goal.multiply(pyaraas.Vector(0,0,-50))
await g.move_to(goal, 0.1)
await g.close()
```
