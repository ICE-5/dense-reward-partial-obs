```python
import pyatk
from pyatk import Transform, Vector
import time

pyatk.init(True)
```

```python
pyatk.set_project_dir("/Users/luor/Documents/projects/RD2/Codes/PyATK-RL/Sample")
w = pyatk.load_workcell("PandaRL")
```

```python
p = w.add_part("peg", "peg0", Transform(500,0,250,0,0,0))
target = w.get_peripheral("insertion_box-0")
target.set_collision_model(True, True)
```

```python
g = w.get_gripper("PdDefault-0")
g.attach(p, False)
p.set_pose(Transform(560,50,350,0,-0,0), 0, g)
g.attach(p, True)
p.create_vel_controlled_tool(True)
```

```python
for i in range(0,1000):
    state = p.set_vel(Vector(0,0,100),Vector(0,0,0), 0.05)
    # print(i, "force", state[1])
    # print("pos", state[3].position)
    if state[0] != "valid":
        print(state[0])
        break
    time.sleep(0.05)
    
```
